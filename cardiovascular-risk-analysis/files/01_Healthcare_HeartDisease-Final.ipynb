```python
# 1. Environment Setup & Data Ingestion
import pandas as pd
import numpy as np

url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data'
columns = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']

df = pd.read_csv(url, header=None, names=columns)

# 2. Missing Value Imputation & Type Casting
df.replace('?', np.nan, inplace=True)
df['ca'] = pd.to_numeric(df['ca'])
df['thal'] = pd.to_numeric(df['thal'])

df['ca'].fillna(df['ca'].median(), inplace=True)
df['thal'].fillna(df['thal'].median(), inplace=True)

# 3. Categorical Transformation & Domain Mapping
df['Target_Label'] = df['target'].apply(lambda x: 'Heart Disease' if x > 0 else 'Healthy')

age_bins = [0, 44, 54, 64, 100]
age_labels = ['<45', '45-54', '55-64', '65+']
df['Age_Group'] = pd.cut(df['age'], bins=age_bins, labels=age_labels)

# 4. Exporting Processed Dataset for BI Tool
df.to_csv('heart_disease_tableau_ready.csv', index=False)
```
